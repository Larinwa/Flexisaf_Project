#OBJECT DETECTION; FACE DETECTION MADE EASY

Object detection is the process of automatically identifying and locating objects within an image. A common example is face detection, which is widely used in cameras, security systems, and social media applications. In this project, OpenCV’s Haar cascades are used to detect faces in images. Haar cascades are pre-trained classifiers that identify patterns like faces based on edges and textures. To make the process easier, the detection workflow was wrapped into a function, allowing the user to simply upload an image and call the function to perform face detection automatically.


# Steps Taken:

### Step 1: Setup environment
The necessary libraries, OpenCV and NumPy, were installed. OpenCV provides tools for computer vision, including face detection, while NumPy facilitates image data manipulation. Installing these libraries ensures that images can be processed efficiently.

### Step 2: Upload an image
An image was uploaded to serve as input for the detection process. Since the project focuses on static images, this step provides the file to be analyzed. In Colab, files.upload() is used to select and import images easily.

### Step 3: Load image and convert to grayscale
The uploaded image was read using OpenCV and converted to grayscale. Haar cascades perform faster and more accurately on grayscale images because they only require intensity information instead of full color.

### Step 4: Load Haar cascade for face detection
A pre-trained Haar cascade XML file for faces was loaded. This file contains patterns that allow the algorithm to recognize faces within an image. OpenCV offers several cascades for detecting faces, eyes, smiles, and other objects.

### Step 5: Detect faces and draw bounding boxes
The detectMultiScale method was used to locate faces in the grayscale image. Green rectangles were drawn around detected faces to visualize the results, showing exactly where the algorithm identifies faces.

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow
```

```python
# Load Haar cascade for face detection
faceCascade = cv2.CascadeClassifier(cv2.data.haarcascades + "haarcascade_frontalface_default.xml")
```

```python
# Function to detect faces in an image
def detect_faces_from_upload(uploaded):
    """
    uploaded: dictionary from files.upload()
    """
    # Get the uploaded file
    filename = list(uploaded.keys())[0]

    # Read the image
    img = cv2.imread(filename)

    # Convert to grayscale
    imgGray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # Detect faces
    faces = faceCascade.detectMultiScale(imgGray, scaleFactor=1.3, minNeighbors=5)

    # Draw rectangles around faces
    for (x, y, w, h) in faces:
        cv2.rectangle(img, (x, y), (x + w, y + h), (0, 255, 0), 3)

    # Show the result
    cv2_imshow(img)
```

```python
from google.colab import files
uploaded = files.upload()  # this will open a file picker
```

```python
detect_faces_from_upload(uploaded)
```
